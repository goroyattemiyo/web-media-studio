# WMS Colab Companion — Gate 0

IrodoriTTS Studioで実績のある **Colab + Gradio share** 方式を、WMS内iframeへ拡張できるか確認するための接続テストです。

このNotebookでは **yt-dlp / 動画Downloadは実行しません**。
確認するのは Gradio起動、URL受け渡し、iframe表示、ボタン操作、テストファイルDownloadだけです。


```python
#@title STEP 1 — Gate 0を準備 { display-mode: "form" }
WMS_REF = "main" #@param {type:"string"}

import pathlib
import subprocess
import sys
import urllib.request

GRADIO_VERSION = "6.26.0"
ROOT = pathlib.Path("/content/wms-colab-companion-gate0")
ROOT.mkdir(parents=True, exist_ok=True)

print(f"Gradio {GRADIO_VERSION} を準備しています…")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", f"gradio=={GRADIO_VERSION}"], check=True)

base = f"https://raw.githubusercontent.com/goroyattemiyo/web-media-studio/{WMS_REF}/colab_companion"
for name in ["wms_colab_companion_gate0.py", "wms_colab_companion_gate0.css"]:
    target = ROOT / name
    urllib.request.urlretrieve(f"{base}/{name}", target)
    print(f"取得: {target}")

print("準備完了。STEP 2を実行してください。")
```
